```python
# Lấy trang thứ 2 của comment

next_page_token = data_comments["nextPageToken"]

params = {
    "part": "snippet,replies",
    "videoId": "prw-laHI2i8",
    "maxResults": 5,
    "pageToken": next_page_token,
    "key": os.getenv("YOUTUBE_API_KEY")
}

response = requests.get(
    "https://www.googleapis.com/youtube/v3/commentThreads",
    params=params
)

print(response.status_code)

data_comments_page_2 = response.json()

print("Số comment:", len(data_comments_page_2["items"]))
```

```python
comments = []

page_token = None

while True:

    params = {
        "part": "snippet,replies",
        "videoId": "prw-laHI2i8",
        "maxResults": 5,
        "key": os.getenv("YOUTUBE_API_KEY")
    }

    # Nếu có token thì lấy trang tiếp theo
    if page_token:
        params["pageToken"] = page_token

    response = requests.get(
        "https://www.googleapis.com/youtube/v3/commentThreads",
        params=params
    )

    data_comments = response.json()

    print("Đang lấy", len(data_comments["items"]), "comments")

    # Xử lý comments
    for item in data_comments["items"]:

        comment = item["snippet"]["topLevelComment"]

        comment_record = {
            "artist_name": "Sơn Tùng M-TP",
            "channel_id": item["snippet"]["channelId"],
            "video_id": item["snippet"]["videoId"],
            "comment_id": comment["id"],
            "parent_id": None,
            "author_name": comment["snippet"]["authorDisplayName"],
            "comment_text": comment["snippet"]["textOriginal"],
            "published_at": comment["snippet"]["publishedAt"],
            "updated_at": comment["snippet"]["updatedAt"],
            "like_count": int(comment["snippet"].get("likeCount", 0)),
            "reply_count": item["snippet"].get("totalReplyCount", 0),
            "extracted_at": extracted_at,
            "ingestion_date": ingestion_date,
        }

        comments.append(comment_record)

    # Lấy token của trang tiếp theo
    page_token = data_comments.get("nextPageToken")

    # Không còn trang tiếp theo → dừng
    if not page_token:
        break

print("Tổng số comments:", len(comments))
```
